```python
n = int(input())

i = 1
while i <= n:

 
    j = 1
    while j <= n - i:
        print(" ", end="")
        j += 1

    k = 1
    while k <= i:
        print((2*i - 2) + k, end="")
        k += 1

 
    l = 2*i - 2
    while l >= i:
        print(l, end="")
        l -= 1

    print()
    i += 1
```

```python
# Program to print the Fibonacci series till n terms

n = int(input())

num1 = 0
num2 = 1

print(num1, end=" ")
print(num2, end=" ")

for i in range(3, n + 1):
    s = num1 + num2
    print(s, end=" ")

    # SWAP
    num1 = num2
    num2 = s
```

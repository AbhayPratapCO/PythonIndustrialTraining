```python
def binarySearch(arr, k):
    i = 0
    j = len(arr) - 1

    while i <= j:
        mid = (i + j) // 2
        if arr[mid] == k:
            return mid
        elif arr[mid] < k:
            i = mid + 1
        else:
            j = mid - 1
    return -1
```

```python
def bubbleSort(arr):
    n = len(arr)

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

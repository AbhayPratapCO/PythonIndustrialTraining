```python
n = int(input())

i = 0
while i < n:
    j = 0
    while j < n:
        print("*", end = "")
        j += 1

    i += 1
    print()
```

```python
# right triangle pattern
n = int(input())

i = 1
while i <= n:
    j = 1
    while j <= i:
        print("*", end = "")
        j += 1

    i += 1
    print()
```

```python
n = int(input())

i = 1
while i <= n:
    j = 0
    while (j <= n - i + 1):
        print("*", end = "")
        j += 1

    i += 1
    print()
```

```python
n = int(input())

i = 1
while i <= n:
    j = 1
    while j <= n - i:
        print(" ", end="")
        j += 1

    k = 1
    while k <= 2*i - 1:
        print("*", end="")
        k += 1

    print()
    i += 1
```

```python
n = int(input())

i = 0
while i < n:
    j = 0
    while j < i:
        print(" ", end="")
        j += 1

    k = 0
    while k < n - i:
        print("*", end="")
        k += 1

    print()
    i += 1

i = n - 2
while i >= 0:
    j = 0
    while j < i:
        print(" ", end="")
        j += 1

    k = 0
    while k < n - i:
        print("*", end="")
        k += 1

    print()
    i -= 1
```

```python
n = int(input())

i = 1
while i <= n:

    j = 1
    while j <= n - i:
        print(" ", end="")
        j += 1

    k = 0
    while k < i:
        print(i + k, end="")
        k += 1

    l = 1
    while l < i:
        print(i + i - l - 1, end="")
        l += 1

    print()
    i += 1
```
